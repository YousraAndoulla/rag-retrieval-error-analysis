## 1. Load and Inspect the Dataset

```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/synthetic_log_file.csv')
print(df.head())
```

## 2. Filter Refund Policy Interactions

```python
refund_df = df[df['user_intent'] == 'ask_refund_policy'].copy()
```

## 3. Count Interactions in the Selected Segment

```python
print(f"Analysing {len(refund_df)} interactions for the'ask_refund_policy' intent")
```

## 4. Create a Retrieval–Error Contingency Table

```python
contigency_table = pd.crosstab(refund_df['retrieval_status'],refund_df['error_category'])
print(contigency_table)
```

## 5. Calculate Error Distribution by Retrieval Status

```python
error_rate_table = pd.crosstab(refund_df['retrieval_status'],refund_df['error_category'],normalize='index')
print((error_rate_table*100).round(1))
```
